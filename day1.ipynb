```python
# Q5 — Linear Algebra
A = np.array([
    [1, 2],
    [3, 4]
])

# Find:

# determinant
# inverse
# eigenvalues

det= np.linalg.det(A)
print(det)

inverse= np.linalg.inv(A)
print(inverse)

values,eign= np.linalg.eig(A)
print(eign)
```
