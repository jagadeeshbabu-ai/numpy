Project 1 — E-Commerce Sales Analysis

```python
import numpy as np

sales = np.array([
    1200, 4500, 2300, 8000, 1500,
    9200, 3400, 5600, 11000, 2100
])

print(sales)
```

```python
# total sales
print(sales.sum())
```

```python
# average
print(sales.mean())
```

```python
# max sale
print(sales.max())
```

```python
# min
print(sales.min())
```

```python
data=sales[sales > 5000]
data
```

```python
count=np.sum(sales > 5000)
print(count)
```
