Project 5 — ML preprocessing

```python
X = np.array([
    [10, 100],
    [20, 200],
    [30, 300],
    [40, 400]
])

mean = X.mean(axis=0)
std = X.std(axis=0)
print(mean)
print(std)
```

```python
X_scaled = (X - mean) / std
X_scaled
```
